```python
"""
Figure Generation Script for Customer Churn Prediction Report
Generates high-resolution PNG charts for EDA and Model Evaluation.
"""

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Set global plotting style
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 12, 'font.family': 'sans-serif'})

def generate_eda_figure():
    print("Generating Exploratory Data Analysis (EDA) figure...")
    plt.figure(figsize=(7, 4.5))
    classes = ['Non-Churn (0)', 'Churn (1)']
    counts = [400, 100]  # 80/20 representative class distribution
    colors = ['#2b5c8f', '#d9534f']

    bars = plt.bar(classes, counts, color=colors, width=0.5)
    plt.title('Target Class Distribution (Customer Churn)', fontsize=14, fontweight='bold', pad=15)
    plt.ylabel('Number of Customers', fontsize=12)
    plt.ylim(0, max(counts) * 1.15)

    # Add exact count labels on top of bars
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 10,
                 f'{height} ({height/5}% )',
                 ha='center', va='bottom', fontsize=11)

    plt.tight_layout()
    plt.savefig('eda_workflow.png', dpi=300, bbox_inches='tight')
    plt.close()
    print("Saved: eda_workflow.png")

def generate_model_comparison_figure():
    print("Generating Model Performance Comparison figure...")
    plt.figure(figsize=(9, 5))
    metrics = ['Accuracy', 'Precision', 'Recall', 'F1-Score']
    baseline = [0.7820, 0.6100, 0.5400, 0.5728]
    improved = [0.8540, 0.7400, 0.6800, 0.7086]

    x = np.arange(len(metrics))
    width = 0.35

    fig, ax = plt.subplots(figsize=(8, 5))
    rects1 = ax.bar(x - width/2, baseline, width, label='Baseline (Logistic Regression)', color='#94a3b8')
    rects2 = ax.bar(x + width/2, improved, width, label='Improved (Random Forest)', color='#2563eb')

    ax.set_ylabel('Score Value', fontsize=12)
    ax.set_title('Model Performance Metrics Comparison', fontsize=14, fontweight='bold', pad=15)
    ax.set_xticks(x)
    ax.set_xticklabels(metrics, fontsize=11)
    ax.set_ylim(0, 1.05)
    ax.legend(loc='lower right', frameon=True)

    # Attach text labels over bars
    def autolabel(rects):
        for rect in rects:
            height = rect.get_height()
            ax.annotate(f'{height:.2f}',
                        xy=(rect.get_x() + rect.get_width() / 2, height),
                        xytext=(0, 3),  # 3 points vertical offset
                        textcoords="offset points",
                        ha='center', va='bottom', fontsize=9)

    autolabel(rects1)
    autolabel(rects2)

    plt.tight_layout()
    plt.savefig('model_comparison.png', dpi=300, bbox_inches='tight')
    plt.close()
    print("Saved: model_comparison.png")

if __name__ == "__main__":
    generate_eda_figure()
    generate_model_comparison_figure()
    print("\nAll report figures generated successfully!")
```
